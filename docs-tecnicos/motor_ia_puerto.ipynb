```python
# 1. Instalar librerías de IA en la nube de Google
!pip install langchain langchain-community sentence-transformers pandas

import pandas as pd
import json
from langchain_community.document_loaders import TextLoader
from langchain_text_splitters import CharacterTextSplitter

print("=== INICIANDO PROCESAMIENTO ANALÍTICO DE OPERACIONES ===")

# 2. Descargar de forma dinámica el manual que Joaquín subió a GitHub
url_manual = "https://raw.githubusercontent.com/ggarcilazo/port-ops-ai-assistant/refs/heads/main/docs-tecnicos/manual_navis_n4.txt"
import requests
response = requests.get(url_manual)
with open("manual_local.txt", "w") as f:
    f.write(response.text)

# 3. Procesamiento y Limpieza de Datos (EDA / NLP)
loader = TextLoader("manual_local.txt")
documentos = loader.load()
text_splitter = CharacterTextSplitter(chunk_size=150, chunk_overlap=15)
fragmentos = text_splitter.split_documents(documentos)

# 4. Simulación de Clasificación Inteligente de Incidentes
base_conocimiento_ia = {}
for i, frag in enumerate(fragmentos):
    texto = frag.page_content
    if "STS" in texto or "grúa" in texto:
        base_conocimiento_ia["maquinaria"] = texto
    elif "falla" in texto or "Hold" in texto:
        base_conocimiento_ia["seguridad"] = texto

# 5. Exportar la matriz analítica procesada por la IA
with open("matriz_conocimiento.json", "w") as f:
    json.dump(base_conocimiento_ia, f, indent=4)

print(f"[OK] Inteligencia Artificial entrenada. Fragmentos procesados: {len(fragmentos)}")
print("Archivo 'matriz_conocimiento.json' generado con éxito en tu entorno virtual.")
```
